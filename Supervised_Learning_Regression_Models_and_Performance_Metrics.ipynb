```python
#Question 1: What is Simple Linear Regression (SLR)? Explain its purpose.

Answer:
Simple Linear Regression is a method used to understand the relationship between one input variable and one output variable.
Its main purpose is to predict a value and see how one variable affects another using a straight line.
```

```python
#Question 2: What are the key assumptions of Simple Linear Regression?

Answer:
Simple Linear Regression works well when:

The relationship between variables is linear

Errors are independent

Errors have constant variance

Errors are normally distributed

There are no extreme outliers
```

```python
#Question 3: Write the mathematical equation for a simple linear regression model and explain each term.

Answer:
The equation is:

y = mx + c

Where:

y = dependent variable (output)

x = independent variable (input)

m = slope (change in y when x increases by 1)

c = intercept (value of y when x = 0)
```

```python
#Question 4: Provide a real-world example where simple linear regression can be applied.

Answer:
Simple Linear Regression can be used to predict house prices based on house size.
As the size increases, the price usually increases in a linear way.
```

```python
#Question 5: What is the method of least squares in linear regression?

Answer:
The method of least squares finds the best-fit line by minimizing the sum of squared errors between actual values and predicted values.
It helps make predictions as accurate as possible.
```

```python
#Question 6: What is Logistic Regression? How does it differ from Linear Regression?

Answer:
Logistic Regression is used for classification problems, such as yes/no or pass/fail.
Linear Regression predicts continuous values, while Logistic Regression predicts probabilities and categories.
```

```python
#Question 7: Name and briefly describe three common evaluation metrics for regression models.

Answer:

Mean Absolute Error (MAE): Average of absolute prediction errors

Mean Squared Error (MSE): Average of squared errors

Root Mean Squared Error (RMSE): Square root of MSE, easier to understand
```

```python
#Question 8: What is the purpose of the R-squared metric in regression analysis?

Answer:
R-squared shows how well the model explains the data.
A value close to 1 means the model fits the data very well.
```

```python
#Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 6, 8, 10])

# Create and fit model
model = LinearRegression()
model.fit(X, y)

# Print slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
#Question 10: How do you interpret the coefficients in a simple linear regression model?

Answer:
The slope shows how much the output changes when the input increases by one unit.
The intercept shows the output value when the input is zero.
```
