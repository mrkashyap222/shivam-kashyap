```python
Question 1: What is Anomaly Detection? Explain its types with examples.

Answer:

Anomaly Detection is the process of finding data points that are very different from normal data. These unusual points are called anomalies or outliers.

Types of Anomalies:

Point Anomaly

A single data point that is abnormal.

Example: A taxi fare of ₹10,000 when normal fares are around ₹300.

Contextual Anomaly

Data is abnormal only in a specific context (time, season, location).

Example: High electricity usage at midnight (abnormal), but normal during daytime.

Collective Anomaly

A group of data points together is abnormal.

Example: Continuous drop in sensor readings indicating machine failure.
```

```python
Question 3: What are the key components of a Time Series?

Answer:

A Time Series has four main components:

Trend – Long-term increase or decrease

Example: Gradual increase in airline passengers

Seasonality – Repeating pattern over time

Example: Higher sales during festivals

Cyclic – Long-term ups and downs without fixed period

Example: Economic cycles

Residual (Noise) – Random variation

Example: Sudden small spikes due to unexpected events
```

```python
Question 4: What is Stationarity? How to test and transform it?

Answer:

A stationary time series has:

Constant mean

Constant variance

No changing trend over time

How to test stationarity:

ADF Test (Augmented Dickey-Fuller Test)

How to make data stationary:

Differencing

Log transformation

Removing trend or seasonality
```

```python
Question 6: Load AirPassengers, plot & decompose
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load dataset
data = pd.read_csv("AirPassengers.csv")
data['Month'] = pd.to_datetime(data['Month'])
data.set_index('Month', inplace=True)

# Plot original series
plt.figure()
plt.plot(data)
plt.title("Air Passengers")
plt.show()

# Decomposition
decompose = seasonal_decompose(data, model='multiplicative')
decompose.plot()
plt.show()
```

```python
Question 7: Apply Isolation Forest on NYC Taxi Fare
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Sample data
df = pd.DataFrame({
    'fare': [10, 12, 11, 300, 13, 12, 500, 11]
})

# Isolation Forest
model = IsolationForest(contamination=0.25)
df['anomaly'] = model.fit_predict(df[['fare']])

# Plot
plt.scatter(range(len(df)), df['fare'], c=df['anomaly'])
plt.title("Taxi Fare Anomalies")
plt.show()
```

```python
Question 8: Train SARIMA & forecast next 12 months

from statsmodels.tsa.statespace.sarimax import SARIMAX

# SARIMA Model
model = SARIMAX(data, order=(1,1,1), seasonal_order=(1,1,1,12))
results = model.fit()

# Forecast
forecast = results.forecast(12)

# Plot
plt.plot(data, label='Actual')
plt.plot(forecast, label='Forecast')
plt.legend()
plt.show()
```

```python
Question 9: Apply Local Outlier Factor (LOF)
from sklearn.neighbors import LocalOutlierFactor
import numpy as np

X = np.array([[1],[2],[3],[100],[2],[3]])

lof = LocalOutlierFactor(n_neighbors=2)
y_pred = lof.fit_predict(X)

plt.scatter(range(len(X)), X, c=y_pred)
plt.title("LOF Anomaly Detection")
plt.show()
```
